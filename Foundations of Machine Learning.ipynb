### Question:
What is the difference between AI, ML, DL, and Data Science? Provide a brief explanation of each. (Hint: Compare their scope, techniques, and applications for each.)

### Answer:

**1. Artificial Intelligence (AI)**
*   **Scope:** AI is the broadest field, aiming to create intelligent machines that can simulate human intelligence. It encompasses any technique that enables computers to mimic human cognitive functions like learning, problem-solving, perception, and decision-making.
*   **Techniques:** Symbolic AI (expert systems, rule-based systems), search algorithms, logic programming, natural language processing, computer vision, and machine learning.
*   **Applications:** Self-driving cars, virtual assistants (Siri, Alexa), robotics, medical diagnosis, game playing (e.g., AlphaGo).

**2. Machine Learning (ML)**
*   **Scope:** ML is a subset of AI that focuses on enabling systems to learn from data without being explicitly programmed. It's about developing algorithms that can identify patterns in data and make predictions or decisions.
*   **Techniques:** Supervised learning (regression, classification), unsupervised learning (clustering, dimensionality reduction), reinforcement learning (Q-learning, SARSA), decision trees, support vector machines (SVMs), k-nearest neighbors (KNN), neural networks.
*   **Applications:** Spam detection, recommendation systems (Netflix, Amazon), fraud detection, predictive maintenance, stock market prediction.

**3. Deep Learning (DL)**
*   **Scope:** DL is a specialized subset of Machine Learning that uses neural networks with many layers (deep neural networks) to learn complex patterns from large amounts of data. It's particularly effective for unstructured data like images, audio, and text.
*   **Techniques:** Convolutional Neural Networks (CNNs), Recurrent Neural Networks (RNNs), Transformers, Generative Adversarial Networks (GANs), autoencoders.
*   **Applications:** Facial recognition, natural language translation, voice recognition, image generation, medical image analysis.

**4. Data Science**
*   **Scope:** Data Science is an interdisciplinary field that uses scientific methods, processes, algorithms, and systems to extract knowledge and insights from structured and unstructured data. It involves the entire lifecycle of data, from collection and cleaning to analysis and interpretation.
*   **Techniques:** Statistics, probability, data visualization, programming (Python, R), database management, data mining, and often leverages ML/DL algorithms for predictive modeling.
*   **Applications:** Business intelligence, market analysis, customer segmentation, scientific research, epidemiological studies, urban planning.

**In summary:**
*   **AI** is the big umbrella: making machines intelligent.
*   **ML** is a way to achieve AI: machines learn from data.
*   **DL** is a specific, powerful type of ML: using deep neural networks to learn from complex data.
*   **Data Science** is a field that uses all these tools (and more, like statistics and domain expertise) to extract insights and value from data for various applications.

### Question 2: Explain overfitting and underfitting in ML. How can you detect and prevent them?
Hint: Discuss bias-variance tradeoff, cross-validation, and regularization techniques.

### Answer:

### **Overfitting**

*   **Explanation:** Overfitting occurs when a model learns the training data too well, including the noise and outliers, to the extent that it performs poorly on new, unseen data. It essentially memorizes the training examples rather than learning the underlying patterns.
*   **Characteristics:** High variance, low bias. The model is too complex for the given data.
*   **Detection:**
    *   **Performance Gap:** A significant difference between training accuracy (high) and validation/test accuracy (low).
    *   **Learning Curves:** On a learning curve, the training error will be very low, while the validation error will be high and may even increase after a certain point.
*   **Prevention:**
    *   **More Training Data:** Providing more diverse data helps the model generalize better.
    *   **Feature Selection/Engineering:** Reducing the number of features or creating more relevant ones can simplify the model.
    *   **Cross-Validation:** Techniques like k-fold cross-validation help assess model performance on different subsets of the data, providing a more robust estimate of generalization error and indicating overfitting if performance varies widely.
    *   **Regularization:** Adding a penalty term to the loss function (e.g., L1 Lasso, L2 Ridge regularization) discourages overly complex models by penalizing large coefficients.
    *   **Early Stopping:** Stopping the training process before the model starts to memorize the training data (i.e., when validation error starts to increase).
    *   **Simpler Model:** Using a less complex model with fewer parameters.

### **Underfitting**

*   **Explanation:** Underfitting occurs when a model is too simple to capture the underlying patterns in the training data, resulting in poor performance on both training and new data. It fails to learn the relationships between features and the target variable.
*   **Characteristics:** High bias, low variance. The model is too simple for the given data.
*   **Detection:**
    *   **Poor Performance:** Low accuracy/high error on both training and validation/test datasets.
    *   **Learning Curves:** On a learning curve, both training and validation errors will be high and plateau at a similar high level.
*   **Prevention:**
    *   **More Complex Model:** Using a more sophisticated model with more parameters (e.g., more layers in a neural network, more complex decision trees).
    *   **More Features:** Adding more relevant features to help the model identify patterns.
    *   **Reduce Regularization:** Decreasing the strength of regularization if it's too high, as strong regularization can over-simplify the model.
    *   **Longer Training:** Training the model for more epochs can sometimes help it learn more.

### **Bias-Variance Tradeoff**

*   **Concept:** This is a central concept in machine learning that helps explain overfitting and underfitting. It states that as you decrease bias (make the model more complex to capture more patterns), you typically increase variance (make it more sensitive to fluctuations in the training data), and vice-versa.
    *   **High Bias (Underfitting):** The model makes strong assumptions about the data, leading it to miss relevant relations between features and target outputs (oversimplification). It has a high error on both training and test sets.
    *   **High Variance (Overfitting):** The model is too sensitive to the training data, capturing random noise rather than the intended outputs. It performs well on the training set but poorly on the test set.
*   **Goal:** The aim is to find a balance between bias and variance that minimizes the total error on unseen data. This sweet spot represents a model that generalizes well.

By carefully considering these concepts and applying the appropriate detection and prevention techniques, you can build more robust and effective machine learning models.

### Question 3:How would you handle missing values in a dataset? Explain at least three methods with examples.
Hint: Consider deletion, mean/median imputation, and predictive modeling.

### Answer:

Missing values are a common problem in real-world datasets and can significantly impact the performance of machine learning models. Here are three common methods to handle them:

1.  **Deletion:**
    *   **Explanation:** This method involves removing rows (samples) or columns (features) that contain missing values.
    *   **When to use:**
        *   **Row Deletion (Listwise Deletion):** If only a small percentage of rows have missing values, deleting them might be acceptable. However, it can lead to significant data loss if many rows have missing data, potentially introducing bias if the missingness is not completely random.
        *   **Column Deletion:** If a feature has a very high percentage of missing values (e.g., >70-80%), or if it's not considered crucial for the analysis, the entire column can be dropped.
    *   **Example (Python - Pandas):**
        ```python
        import pandas as pd
        import numpy as np

        df = pd.DataFrame({
            'A': [1, 2, np.nan, 4, 5],
            'B': [6, np.nan, 8, 9, 10],
            'C': [11, 12, 13, 14, np.nan]
        })

        # Drop rows with any missing values
        df_rows_dropped = df.dropna()
        print("\nDataFrame after dropping rows:")
        print(df_rows_dropped)

        # Drop columns with any missing values
        df_cols_dropped = df.dropna(axis=1)
        print("\nDataFrame after dropping columns:")
        print(df_cols_dropped)
        ```

2.  **Imputation (Mean, Median, Mode):**
    *   **Explanation:** This involves filling missing values with a calculated statistic (mean, median, or mode) from the non-missing values of the respective feature. This helps retain data that would otherwise be lost through deletion.
    *   **When to use:**
        *   **Mean Imputation:** Suitable for numerical data when the data distribution is symmetrical (not skewed) and outliers are not a major concern.
        *   **Median Imputation:** Preferred for numerical data when the data is skewed or contains outliers, as the median is less sensitive to extreme values.
        *   **Mode Imputation:** Primarily used for categorical data, but can also be used for numerical data if a single value occurs most frequently.
    *   **Example (Python - Pandas/Scikit-learn):**
        ```python
        import pandas as pd
        import numpy as np
        from sklearn.impute import SimpleImputer

        df = pd.DataFrame({
            'Numeric_Skewed': [10, 20, np.nan, 30, 1000],
            'Numeric_Normal': [1.1, 2.2, np.nan, 3.3, 4.4],
            'Categorical': ['A', 'B', 'A', np.nan, 'C']
        })

        # Median Imputation for 'Numeric_Skewed'
        median_imputer = SimpleImputer(strategy='median')
        df['Numeric_Skewed_Imputed'] = median_imputer.fit_transform(df[['Numeric_Skewed']])

        # Mean Imputation for 'Numeric_Normal'
        mean_imputer = SimpleImputer(strategy='mean')
        df['Numeric_Normal_Imputed'] = mean_imputer.fit_transform(df[['Numeric_Normal']])

        # Mode Imputation for 'Categorical'
        mode_imputer = SimpleImputer(strategy='most_frequent')
        df['Categorical_Imputed'] = mode_imputer.fit_transform(df[['Categorical']])

        print("\nDataFrame after Mean/Median/Mode Imputation:")
        print(df)
        ```

3.  **Predictive Modeling (e.g., K-Nearest Neighbors Imputation, Regression Imputation):**
    *   **Explanation:** More sophisticated methods where missing values are predicted using other features in the dataset. This treats the feature with missing values as a target variable and uses other features as predictors.
    *   **When to use:** When the missingness is not completely random and there are discernible relationships between the features. It's generally more accurate than simple imputation but computationally more expensive.
    *   **Example (Python - Scikit-learn's IterativeImputer (MICE) or KNNImputer):**
        ```python
        import pandas as pd
        import numpy as np
        from sklearn.impute import KNNImputer
        from sklearn.experimental import enable_iterative_imputer # Required for IterativeImputer
        from sklearn.impute import IterativeImputer

        df = pd.DataFrame({
            'Feature1': [1, 2, np.nan, 4, 5],
            'Feature2': [10, np.nan, 30, 40, 50],
            'Feature3': [100, 200, 300, np.nan, 500]
        })

        # KNN Imputation
        knn_imputer = KNNImputer(n_neighbors=2)
        df_knn_imputed = pd.DataFrame(knn_imputer.fit_transform(df), columns=df.columns)
        print("\nDataFrame after KNN Imputation:")
        print(df_knn_imputed)

        # Iterative Imputation (MICE)
        iterative_imputer = IterativeImputer(max_iter=10, random_state=0)
        df_iter_imputed = pd.DataFrame(iterative_imputer.fit_transform(df), columns=df.columns)
        print("\nDataFrame after Iterative Imputation (MICE):")
        print(df_iter_imputed)
        ```

### Question 4:What is an imbalanced dataset? Describe two techniques to handle it (theoretical + practical).
Hint: Discuss SMOTE, Random Under/Oversampling, and class weights in models.

### Answer:

An **imbalanced dataset** is a dataset where the number of observations belonging to one class (the majority class) is significantly higher than the number of observations belonging to other classes (the minority class). This imbalance can cause machine learning models to perform poorly on the minority class because they tend to be biased towards the majority class, which dominates the training process.

Here are two techniques to handle imbalanced datasets:

1.  **Resampling Techniques (Random Over-sampling and Under-sampling):**
    *   **Theoretical:**
        *   **Random Over-sampling:** Involves randomly duplicating samples from the minority class to increase its representation in the dataset. This balances the class distribution but can lead to overfitting as the model might learn specific duplicated instances too well.
        *   **Random Under-sampling:** Involves randomly removing samples from the majority class to reduce its representation. This can lead to loss of potentially valuable information from the majority class if too many samples are removed.
    *   **Practical (Python - `imblearn` library):**
        ```python
        from collections import Counter
        from sklearn.datasets import make_classification
        from imblearn.over_sampling import RandomOverSampler
        from imblearn.under_sampling import RandomUnderSampler

        # Create a highly imbalanced dataset
        X, y = make_classification(n_samples=1000, n_features=2, n_informative=2,
                                   n_redundant=0, n_repeated=0, n_classes=2,
                                   n_clusters_per_class=1, weights=[0.95, 0.05],
                                   flip_y=0, random_state=42)
        print(f"Original dataset shape: {Counter(y)}")

        # Random Over-sampling
        ros = RandomOverSampler(random_state=42)
        X_resampled_ros, y_resampled_ros = ros.fit_resample(X, y)
        print(f"Random Over-sampled dataset shape: {Counter(y_resampled_ros)}")

        # Random Under-sampling
        rus = RandomUnderSampler(random_state=42)
        X_resampled_rus, y_resampled_rus = rus.fit_resample(X, y)
        print(f"Random Under-sampled dataset shape: {Counter(y_resampled_rus)}")
        ```

2.  **SMOTE (Synthetic Minority Over-sampling Technique):**
    *   **Theoretical:** SMOTE is an advanced over-sampling technique that addresses the overfitting issue of simple random over-sampling. Instead of just duplicating existing minority class samples, SMOTE creates *synthetic* samples. It works by selecting a minority class instance and then finding its k-nearest neighbors. New synthetic instances are generated along the line segments connecting the original instance to its neighbors.
    *   **Practical (Python - `imblearn` library):**
        ```python
        from collections import Counter
        from sklearn.datasets import make_classification
        from imblearn.over_sampling import SMOTE

        # Create a highly imbalanced dataset
        X, y = make_classification(n_samples=1000, n_features=2, n_informative=2,
                                   n_redundant=0, n_repeated=0, n_classes=2,
                                   n_clusters_per_class=1, weights=[0.95, 0.05],
                                   flip_y=0, random_state=42)
        print(f"Original dataset shape: {Counter(y)}")

        # Apply SMOTE
        smote = SMOTE(random_state=42)
        X_resampled_smote, y_resampled_smote = smote.fit_resample(X, y)
        print(f"SMOTE resampled dataset shape: {Counter(y_resampled_smote)}")
        ```

    *   **Alternative: Class Weights in Models:** Many algorithms (e.g., Logistic Regression, Support Vector Machines, Tree-based models like RandomForestClassifier, XGBoost) allow you to assign different weights to classes during training. This effectively tells the model to pay more attention to the minority class errors. For instance, in scikit-learn, you can often set `class_weight='balanced'` to automatically adjust weights inversely proportional to class frequencies, or provide a custom dictionary of weights.

### Question 5: Why is feature scaling important in ML? Compare Min-Max scaling and Standardization.
Hint: Explain impact on distance-based algorithms (e.g., KNN, SVM) and gradient descent.

### Answer:

**Why Feature Scaling is Important in ML:**

Feature scaling is a data pre-processing step that transforms the range of features in a dataset. It is crucial for several reasons:

1.  **Impact on Distance-Based Algorithms (e.g., K-Nearest Neighbors, Support Vector Machines):**
    *   Algorithms like KNN, SVM, and K-Means clustering calculate the distances between data points. If features are on different scales (e.g., one feature ranges from 0-1000 and another from 0-1), the feature with the larger range will numerically dominate the distance calculation. This can lead to the algorithm being unfairly biased towards the larger-scaled feature, ignoring the contributions of smaller-scaled but potentially important features.
    *   Scaling ensures that all features contribute proportionally to the distance metrics.

2.  **Impact on Gradient Descent Based Algorithms:**
    *   Algorithms that rely on gradient descent optimization (e.g., Logistic Regression, Neural Networks, Gradient Boosting Machines) converge much faster and more stably when features are scaled.
    *   Without scaling, the cost function's contour plots can become elongated and narrow. The gradient descent algorithm might struggle to find the minimum efficiently, oscillating back and forth across the narrow dimensions and taking many more iterations to converge, or even failing to converge at all.
    *   Scaling creates a more spherical cost function surface, allowing gradient descent to take more direct paths to the minimum.

3.  **Regularization:** Many regularization techniques (like L1 and L2 regularization) penalize large coefficients. If features are not scaled, features with larger ranges might end up with smaller coefficients simply because of their scale, rather than their true importance. Scaling ensures that regularization penalties are applied fairly across all features.

**Comparison of Min-Max Scaling and Standardization:**

1.  **Min-Max Scaling (Normalization):**
    *   **Explanation:** This technique rescales features to a fixed range, typically between 0 and 1. The original values are transformed based on the minimum and maximum values of the feature.
    *   **Formula:** `X_scaled = (X - X_min) / (X_max - X_min)`
    *   **When to Prefer:**
        *   When you need features to be within a specific, predefined range (e.g., for image processing where pixel intensities are 0-255, you might scale to 0-1).
        *   Algorithms that are sensitive to the magnitude of values but not necessarily their distribution (e.g., neural networks with sigmoid or tanh activation functions, or algorithms that expect input in a specific range).
    *   **Sensitivity to Outliers:** Highly sensitive to outliers. A single extreme outlier can significantly compress the range of the rest of the data points, making them look very similar after scaling.

2.  **Standardization (Z-score Normalization):**
    *   **Explanation:** This technique transforms features to have a mean of 0 and a standard deviation of 1. It centers the data around the mean and scales it by its standard deviation.
    *   **Formula:** `X_scaled = (X - μ) / σ` (where μ is the mean and σ is the standard deviation)
    *   **When to Prefer:**
        *   Algorithms that assume features follow a Gaussian (normal) distribution (e.g., Linear Regression, Logistic Regression, Linear Discriminant Analysis, Gaussian Naive Bayes).
        *   Algorithms that use distance metrics (KNN, SVM) where outliers are a concern, as standardization handles them better than Min-Max scaling by not bounding the data to a specific range.
        *   When you don't know the exact distribution of your data.
    *   **Sensitivity to Outliers:** Less sensitive to outliers than Min-Max scaling because it uses the mean and standard deviation, which are influenced by outliers, but the scaling doesn't compress the entire range as drastically. Outliers will still be present, just standardized.

**Key Difference:** Min-Max scaling preserves the original distribution's shape but bounds the values to a specific range. Standardization changes the distribution's mean and standard deviation, making it suitable for algorithms that assume a normal distribution or benefit from centered data, and it's generally more robust to outliers.

### Question 6: Compare Label Encoding and One-Hot Encoding. When would you prefer one over the other?
Hint: Consider categorical variables with ordinal vs. nominal relationships.

### Answer:

Label Encoding and One-Hot Encoding are two common techniques used to convert categorical data into numerical format, which is required by most machine learning algorithms.

1.  **Label Encoding:**
    *   **Explanation:** Assigns a unique integer to each category in a categorical feature. For example, if a feature 'Color' has categories 'Red', 'Green', 'Blue', Label Encoding might transform them to 0, 1, 2 respectively.
    *   **Example:**
        ```python
        from sklearn.preprocessing import LabelEncoder
        import pandas as pd

        data = pd.DataFrame({'Size': ['Small', 'Medium', 'Large', 'Small']})
        le = LabelEncoder()
        data['Size_Encoded'] = le.fit_transform(data['Size'])
        print("\nLabel Encoded Data (Size):")
        print(data)
        # Output: Small->1, Medium->0, Large->2 (order depends on alphabetical or fit order)
        # Note: Scikit-learn's LabelEncoder often assigns based on alphabetical order of unique values
        ```
    *   **When to Prefer (Ordinal Relationships):**
        *   **Ordinal Categorical Variables:** This method is suitable when the categorical variable has an **ordinal** relationship, meaning there is an inherent order or ranking among its categories (e.g., 'Small', 'Medium', 'Large'; 'Low', 'Medium', 'High'). In such cases, the numerical representation naturally reflects this order.
        *   **Tree-based Algorithms:** Some tree-based algorithms (like Decision Trees, Random Forests, Gradient Boosting) can sometimes handle label-encoded features even if they are nominal because they split nodes based on thresholds, and the exact numerical relationship might not mislead them as much as it would linear models.
    *   **Disadvantage:** When used with **nominal** categorical variables (categories without any inherent order, like 'City', 'Color', 'Gender'), Label Encoding introduces an artificial ordinal relationship. For instance, if 'Red' is 0, 'Green' is 1, and 'Blue' is 2, a model might incorrectly infer that 'Blue' is 'greater' or 'more important' than 'Red', which is not true. This can severely mislead linear models (e.g., Linear Regression, Logistic Regression) or distance-based algorithms (e.g., KNN, SVM).

2.  **One-Hot Encoding:**
    *   **Explanation:** Creates new binary (dummy) features for each unique category in the original categorical feature. For each original category, a new column is added, and it contains 1 if the observation belongs to that category, and 0 otherwise. This avoids introducing any artificial ordinal relationships.
    *   **Example:**
        ```python
        from sklearn.preprocessing import OneHotEncoder
        import pandas as pd

        data = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})
        ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
        color_encoded = ohe.fit_transform(data[['Color']])
        color_df = pd.DataFrame(color_encoded, columns=ohe.get_feature_names_out(['Color']))
        data_ohe = pd.concat([data, color_df], axis=1)
        print("\nOne-Hot Encoded Data (Color):")
        print(data_ohe)
        ```
    *   **When to Prefer (Nominal Relationships):**
        *   **Nominal Categorical Variables:** This is the preferred method when the categorical variable has a **nominal** relationship (categories without any inherent order, e.g., 'City', 'Color', 'Gender', 'Marital Status'). It prevents the model from assuming an incorrect ordinal relationship between categories.
        *   **Linear Models and Distance-Based Algorithms:** Essential for algorithms that are sensitive to the magnitude and relative ordering of numerical features, such as Linear Regression, Logistic Regression, SVMs, and KNN. These models would be misled by Label Encoding on nominal data.
    *   **Disadvantage:**
        *   **Curse of Dimensionality:** If a categorical feature has many unique categories, One-Hot Encoding can create a large number of new features, leading to a high-dimensional dataset. This can increase computational cost and potentially lead to the 'curse of dimensionality' problems.
        *   **Multicollinearity (Dummy Variable Trap):** If all dummy variables for a feature are kept (e.g., 'Color_Red', 'Color_Green', 'Color_Blue'), one column can be perfectly predicted from the others (e.g., if 'Color_Red' and 'Color_Green' are 0, then 'Color_Blue' must be 1). This multicollinearity can cause issues in some regression models. It's often recommended to drop one of the dummy variables.

In summary, choose **Label Encoding for ordinal categorical variables** or when using algorithms less sensitive to numerical order (like some tree-based models). Choose **One-Hot Encoding for nominal categorical variables** and for algorithms sensitive to numerical relationships (like linear models, SVMs, KNN) to avoid introducing misleading orderings.

a) Compare the survival rates based on passenger class (Pclass). Which class had the highest survival rate, and why do you think that happened?

````python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Titanic dataset
titanic_url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df_titanic = pd.read_csv(titanic_url)
print("Titanic dataset loaded successfully.")

# --- Part a) Survival Rates by Pclass ---
# Calculate survival rate for each 'Pclass'
survival_rate_pclass = df_titanic.groupby('Pclass')['Survived'].mean()

# Print the survival rates
print("\nSurvival Rates by Passenger Class:")
print(survival_rate_pclass)

# Identify and print the 'Pclass' with the highest survival rate
highest_survival_pclass = survival_rate_pclass.idxmax()
highest_survival_rate = survival_rate_pclass.max()
print(f"\nPassenger Class with the Highest Survival Rate: Pclass {highest_survival_pclass} (Survival Rate: {highest_survival_rate:.2f})")

# Create a bar chart for Pclass survival
plt.figure(figsize=(8, 5))
survival_rate_pclass.plot(kind='bar', color=['skyblue', 'lightcoral', 'lightgreen'])
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# --- Part b) Survival Rates by Age Group ---
# Handle missing 'Age' values by imputing with the median
median_age = df_titanic['Age'].median()
df_titanic['Age'] = df_titanic['Age'].fillna(median_age)
print(f"\nMissing 'Age' values imputed with median: {median_age:.2f}.")

# Create 'Age_Group' ('children' < 18, 'adults' >= 18)
def create_age_group(age):
    if age < 18:
        return 'children'
    else:
        return 'adults'

df_titanic['Age_Group'] = df_titanic['Age'].apply(create_age_group)
print("'Age_Group' column created.")

# Compare and visualize survival rates for these age groups
survival_rate_age_group = df_titanic.groupby('Age_Group')['Survived'].mean().sort_values(ascending=False)

# Print the survival rates by age group
print("\nSurvival Rates by Age Group:")
print(survival_rate_age_group)

# Create a bar chart for Age Group survival
plt.figure(figsize=(7, 5))
survival_rate_age_group.plot(kind='bar', color=['lightcoral', 'skyblue'])
plt.title('Survival Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Survival Rate')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````
