## Answer Question 1: DDL, DML, DQL Differences

### Subtask:
Explain the fundamental differences between DDL, DML, and DQL commands in SQL, providing one example for each type of command.


### 1. Data Definition Language (DDL)

**Purpose:** DDL commands are used to define, modify, or delete database structures. They deal with the schema of the database, not the data within the schema. These commands affect the database's structure, like creating tables, indexes, or views.

**Key Characteristics:**
*   They implicitly commit the current transaction, meaning changes made by DDL commands cannot be rolled back.
*   They define the database schema.

**Example:**

```sql
CREATE TABLE Employees (
    EmployeeID INT PRIMARY KEY,
    FirstName VARCHAR(50),
    LastName VARCHAR(50),
    DepartmentID INT
);
```

This `CREATE TABLE` command defines a new table named `Employees` with specified columns and their data types, along with a primary key constraint.

### 2. Data Manipulation Language (DML)

**Purpose:** DML commands are used for managing data within schema objects. These commands allow users to retrieve, insert, delete, and update data in database tables.

**Key Characteristics:**
*   Changes made by DML commands can be rolled back.
*   They operate on the data itself, not the database structure.

**Example:**

```sql
INSERT INTO Employees (EmployeeID, FirstName, LastName, DepartmentID)
VALUES (1, 'John', 'Doe', 101);
```

This `INSERT INTO` command adds a new row of data into the `Employees` table.

### 3. Data Query Language (DQL)

**Purpose:** DQL commands are used for retrieving data from the database. The most common DQL command is `SELECT`.

**Key Characteristics:**
*   They do not modify the database or its structure.
*   They are used to fetch data based on specified criteria.

**Example:**

```sql
SELECT FirstName, LastName, DepartmentID
FROM Employees
WHERE DepartmentID = 101;
```

This `SELECT` command retrieves the `FirstName`, `LastName`, and `DepartmentID` for all employees whose `DepartmentID` is 101.

## Answer Question 2: SQL Constraints

### Subtask:
Explain the purpose of SQL constraints, name and describe three common types, and provide a simple scenario where each would be useful.


### Purpose of SQL Constraints

SQL constraints are rules enforced on data columns in a table. They are used to limit the type of data that can go into a table, ensuring the accuracy and reliability of the data. Constraints ensure data integrity by preventing invalid data from being inserted, updated, or deleted.

### Three Common Types of SQL Constraints

1.  **PRIMARY KEY**
    *   **Description**: A PRIMARY KEY constraint uniquely identifies each record in a table. Primary keys must contain unique values, and they cannot contain NULL values. A table can have only one primary key, which can consist of single or multiple columns.
    *   **Scenario**: In a `Customers` table, a `CustomerID` column would be an ideal primary key to uniquely identify each customer.
    *   **SQL Example**:
        ```sql
        CREATE TABLE Customers (
            CustomerID INT PRIMARY KEY,
            FirstName VARCHAR(50),
            LastName VARCHAR(50),
            Email VARCHAR(100)
        );
        ```

2.  **FOREIGN KEY**
    *   **Description**: A FOREIGN KEY is a field (or collection of fields) in one table that refers to the PRIMARY KEY in another table. It is used to link two tables together. The purpose of the foreign key is to maintain referential integrity, which means that the values in the foreign key column must exist in the primary key column of the referenced table.
    *   **Scenario**: In an `Orders` table, an `CustomerID` column could be a foreign key referencing the `CustomerID` in the `Customers` table. This ensures that every order is linked to a valid, existing customer.
    *   **SQL Example**:
        ```sql
        CREATE TABLE Orders (
            OrderID INT PRIMARY KEY,
            CustomerID INT,
            OrderDate DATE,
            Amount DECIMAL(10, 2),
            FOREIGN KEY (CustomerID) REFERENCES Customers(CustomerID)
        );
        ```

3.  **NOT NULL**
    *   **Description**: The NOT NULL constraint enforces that a column cannot accept NULL values. This ensures that certain columns always contain data, which is crucial for important information.
    *   **Scenario**: In a `Products` table, the `ProductName` column should not be empty. Applying a NOT NULL constraint to `ProductName` ensures that every product has a name.
    *   **SQL Example**:
        ```sql
        CREATE TABLE Products (
            ProductID INT PRIMARY KEY,
            ProductName VARCHAR(255) NOT NULL,
            Price DECIMAL(10, 2)
        );
        ```

## Answer Question 3: LIMIT and OFFSET

### Subtask:
Explain the difference between LIMIT and OFFSET clauses in SQL and demonstrate how to use them together to retrieve the third page of results, assuming each page has 10 records.


### 1. The `LIMIT` Clause

The `LIMIT` clause in SQL is used to restrict the number of rows returned by a query. It specifies the maximum number of rows that the result set should contain. This is particularly useful for controlling the size of the output, for example, when you only need to fetch the top N records or when implementing pagination where you want to show a fixed number of items per page.

**Syntax:**

```sql
SELECT column1, column2, ...
FROM table_name
LIMIT row_count;
```

**Example:**
To retrieve only the first 5 records from a table named `Products`:

```sql
SELECT ProductName, Price
FROM Products
LIMIT 5;
```

### 2. The `OFFSET` Clause

The `OFFSET` clause in SQL is used to skip a specified number of rows before beginning to return the rows from the query. It is primarily used in conjunction with `LIMIT` for pagination, allowing you to retrieve a specific "page" of results by skipping previous pages.

**Syntax:**

```sql
SELECT column1, column2, ...
FROM table_name
OFFSET start_row;
```

**Example:**
To skip the first 5 records and then return the rest from a table named `Customers`:

```sql
SELECT CustomerName, City
FROM Customers
OFFSET 5;
```

### 3. Difference Between `LIMIT` and `OFFSET`

While both `LIMIT` and `OFFSET` are used for controlling the result set of a query, their roles are distinct:

*   **`LIMIT`**: Specifies the *maximum number of rows to return* from the result set. It defines the "size" of your result.
*   **`OFFSET`**: Specifies the *number of rows to skip* from the beginning of the result set before `LIMIT` is applied. It defines the "starting point" of your result.

In essence, `OFFSET` moves the starting window for data retrieval, and `LIMIT` determines the size of that window. They are most powerful when used together for pagination.

### 4. Demonstrating `LIMIT` and `OFFSET` for Pagination

To retrieve the third page of results, assuming each page has 10 records, we need to calculate the correct `OFFSET`.

*   **Page 1:** Records 1-10 (OFFSET 0, LIMIT 10)
*   **Page 2:** Records 11-20 (OFFSET 10, LIMIT 10)
*   **Page 3:** Records 21-30 (OFFSET 20, LIMIT 10)

The general formula for `OFFSET` when retrieving the Nth page with `PageSize` records per page is: `OFFSET (N - 1) * PageSize`.

For the third page (N=3) and 10 records per page (PageSize=10):
`OFFSET (3 - 1) * 10 = 2 * 10 = 20`.

Therefore, the SQL query would look like this:

```sql
SELECT *
FROM YourTableName
ORDER BY YourSortingColumn
LIMIT 10
OFFSET 20;
```

**Explanation:**
*   `LIMIT 10`: This specifies that we want to retrieve 10 records (the size of one page).
*   `OFFSET 20`: This tells the database to skip the first 20 records (the first two pages) before applying the `LIMIT`. This effectively starts our result set from the 21st record, giving us records 21-30, which constitute the third page.

## Answer Question 4: Common Table Expression (CTE)

### Subtask:
Explain what a Common Table Expression (CTE) is in SQL, its main benefits, and provide a simple SQL example demonstrating its usage.


### What is a Common Table Expression (CTE)?

A Common Table Expression (CTE) is a temporary, named result set that you can reference within a single SQL statement (SELECT, INSERT, UPDATE, or DELETE). It's defined using the `WITH` clause, and it provides a way to break down complex queries into simpler, more readable, and manageable logical steps. A CTE is not stored as an object in the database; it exists only for the duration of the query's execution.

### Main Benefits of using CTEs:

1.  **Readability and Maintainability:** CTEs improve the readability of complex queries by breaking them into logical, named sections. This makes it easier to understand the query's flow and logic, especially when dealing with multiple joins or subqueries.
2.  **Modularity:** They allow you to define a subquery once and reference it multiple times within the main query, promoting code reuse within a single statement.
3.  **Recursion:** CTEs are essential for performing recursive queries, such as traversing hierarchical or graph-like data (e.g., organizational charts or bill of materials).
4.  **Simplification of Complex Logic:** They can simplify complex `JOIN` operations and improve query organization, making debugging easier.
5.  **Alternative to Subqueries:** While similar to derived tables (subqueries in the `FROM` clause), CTEs are generally more readable and can be referenced multiple times.

### Simple SQL Example Demonstrating CTE Usage:

Let's imagine we have a table called `Employees` and we want to find all employees who earn more than the average salary of their department. We can use a CTE to first calculate the average salary per department and then use that result in our main query.

```sql
-- Define a CTE named 'DepartmentAverageSalary'
WITH DepartmentAverageSalary AS (
    SELECT
        DepartmentID,
        AVG(Salary) AS AverageSalary
    FROM
        Employees
    GROUP BY
        DepartmentID
)
-- Main query that references the CTE
SELECT
    e.EmployeeName,
    e.Salary,
    e.DepartmentID,
    das.AverageSalary
FROM
    Employees e
JOIN
    DepartmentAverageSalary das ON e.DepartmentID = das.DepartmentID
WHERE
    e.Salary > das.AverageSalary;
```

**Explanation of the example:**

1.  **`WITH DepartmentAverageSalary AS (...)`**: This line defines the CTE named `DepartmentAverageSalary`. The parentheses enclose the query that calculates the average salary for each `DepartmentID`.
2.  **`SELECT DepartmentID, AVG(Salary) AS AverageSalary FROM Employees GROUP BY DepartmentID`**: This is the query inside the CTE. It groups the `Employees` table by `DepartmentID` and calculates the `AverageSalary` for each department.
3.  **`SELECT e.EmployeeName, ... WHERE e.Salary > das.AverageSalary`**: This is the main query. It joins the original `Employees` table (`e`) with our newly defined CTE `DepartmentAverageSalary` (`das`) on `DepartmentID`. The `WHERE` clause then filters the results to show only those employees whose `Salary` is greater than the `AverageSalary` of their respective department, which was calculated in the CTE.

## Answer Question 5: SQL Normalization

### Subtask:
Describe the concept of SQL Normalization and its primary goals. Briefly explain the first three normal forms (1NF, 2NF, 3NF).


### SQL Normalization

SQL Normalization is a systematic process of organizing the columns and tables in a relational database to minimize data redundancy and improve data integrity. The primary goals of normalization are:

1.  **Eliminate Redundant Data**: Reduce the amount of duplicate data stored in the database, saving storage space and preventing inconsistencies.
2.  **Ensure Data Integrity**: Maintain the accuracy and consistency of data across the database, preventing anomalies during data insertion, deletion, or update.
3.  **Improve Data Efficiency**: Make the database structure more efficient for queries and updates.
4.  **Simplify Database Design**: Create a more logical and understandable database schema.

### Normal Forms

Database normalization is categorized into various "normal forms," with the first three (1NF, 2NF, 3NF) being the most commonly applied:

#### 1. First Normal Form (1NF)

A table is in 1NF if it meets the following conditions:

*   **Atomic Values**: Each column contains atomic (indivisible) values, meaning there are no repeating groups or arrays within a single column.
*   **Unique Rows**: Each row in the table is unique, typically ensured by a primary key.

**Example**: Instead of having a column `PhoneNumbers` with multiple phone numbers, you would have separate rows or a linked table for each phone number.

#### 2. Second Normal Form (2NF)

A table is in 2NF if it meets the following conditions:

*   It is in **1NF**.
*   **No Partial Dependency**: All non-key attributes are fully functionally dependent on the entire primary key. This means that no non-key attribute should depend on only a part of a composite primary key.

**Example**: If a table has `(OrderID, ProductID)` as a composite primary key and `ProductName` as a non-key attribute, `ProductName` should not depend only on `ProductID` (a part of the primary key). If it does, `ProductName` should be moved to a separate `Products` table.

#### 3. Third Normal Form (3NF)

A table is in 3NF if it meets the following conditions:

*   It is in **2NF**.
*   **No Transitive Dependency**: There are no transitive functional dependencies for non-key attributes. This means that a non-key attribute should not be dependent on another non-key attribute.

**Example**: If a `Customers` table has `(CustomerID, CustomerName, City, ZipCode)` and `City` determines `ZipCode`, then `ZipCode` is transitively dependent on `CustomerID` via `City`. To achieve 3NF, `City` and `ZipCode` would be moved to a separate `Cities` table, with `City` being a foreign key in the `Customers` table.

## Question 7: Generate Customer Order Report

### Subtask:
Write a SQL query to generate a report showing 'CustomerName', 'Email', and the 'TotalNumberofOrders' for each customer in 'ECommerceDB'. Include customers who have not placed any orders (displaying 0 for their total orders) and order the results by 'CustomerName'.


### Reasoning for the SQL Query

To generate the requested report, I will construct a SQL query that performs the following steps:

1.  **Select Customer Information**: I need `CustomerName` and `Email` from the `Customers` table.
2.  **Count Orders**: I need to count the number of orders for each customer from the `Orders` table.
3.  **Handle Missing Orders**: Since the requirement states to include customers who have not placed any orders and display 0 for their total, a `LEFT JOIN` from `Customers` to `Orders` is essential. This ensures all customers are listed, even if they have no matching orders.
4.  **Group Results**: To count orders per customer, I'll `GROUP BY` `CustomerID`, `CustomerName`, and `Email`.
5.  **Aggregate Count**: The `COUNT(Orders.OrderID)` function will count the orders. Since `LEFT JOIN` can result in `NULL` values for `OrderID` for customers without orders, `COUNT(Orders.OrderID)` will correctly return 0 for those customers (as `COUNT` only counts non-NULL values). If I were to use `COUNT(*)`, it would count the row itself, even if `Orders.OrderID` is null, so `COUNT(Orders.OrderID)` or `COUNT(DISTINCT Orders.OrderID)` is appropriate here.
6.  **Order Results**: Finally, the results will be ordered alphabetically by `CustomerName`.

**Reasoning**:
Now that the reasoning for the SQL query is established, I will execute the SQL query using the `cursor` object to retrieve the customer order report. Then, I will fetch and display the results.



````python
sql_query = '''
SELECT
    c.CustomerName,
    c.Email,
    COUNT(o.OrderID) AS TotalNumberofOrders
FROM
    Customers c
LEFT JOIN
    Orders o ON c.CustomerID = o.CustomerID
GROUP BY
    c.CustomerID, c.CustomerName, c.Email
ORDER BY
    c.CustomerName;
'''

cursor.execute(sql_query)
report_results = cursor.fetchall()

# Get column names for better display
column_names = [description[0] for description in cursor.description]

print("\nCustomer Order Report:")
print(column_names)
for row in report_results:
    print(row)
````

## Question 8: Retrieve Product Information with Category

### Subtask:
Write a SQL query to display the 'ProductName', 'Price', 'StockQuantity', and 'CategoryName' for all products in 'ECommerceDB'. Order the results by 'CategoryName' and then 'ProductName' alphabetically.


### Reasoning for the SQL Query

To fulfill the subtask, I need to retrieve product details and their corresponding category names. This requires joining information from two tables: `Products` and `Categories`.

1.  **Select Columns**: I will select `ProductName`, `Price`, and `StockQuantity` from the `Products` table (`p`) and `CategoryName` from the `Categories` table (`c`).
2.  **Join Tables**: An `INNER JOIN` will be used to connect the `Products` table and the `Categories` table. This is because we only want to display products that have an assigned category, and `INNER JOIN` ensures that only rows with matching `CategoryID` in both tables are included in the result set.
3.  **Join Condition**: The join will be performed on `p.CategoryID = c.CategoryID`, linking products to their respective categories.
4.  **Order Results**: The final result set needs to be ordered. First, it will be sorted by `CategoryName` in ascending alphabetical order. Then, for products within the same category, they will be further sorted by `ProductName` in ascending alphabetical order, as specified in the subtask.

## Question 9: Top 2 Most Expensive Products per Category using CTE and Window Function

### Subtask:
Write a SQL query that uses a Common Table Expression (CTE) and a Window Function (specifically ROW_NUMBER() or RANK()) to display the 'CategoryName', 'ProductName', and 'Price' for the top 2 most expensive products in each 'CategoryName' from 'ECommerceDB'.


### Reasoning for the SQL Query

To identify the top 2 most expensive products within each category, I will construct a SQL query utilizing a Common Table Expression (CTE) and a window function, following the specified instructions:

1.  **Join `Products` and `Categories`**: First, I need to combine information from the `Products` table (for `ProductName` and `Price`) and the `Categories` table (for `CategoryName`). An `INNER JOIN` on `CategoryID` will be used for this.

2.  **Define a CTE for Ranking**: I will define a CTE to calculate the rank of each product within its respective category based on its price. This involves:
    *   Selecting `CategoryName`, `ProductName`, and `Price`.
    *   Using the `ROW_NUMBER()` window function. `ROW_NUMBER()` assigns a unique sequential integer to each row within a partition. It's suitable here because if two products have the same price, `ROW_NUMBER()` will still assign them different ranks, which is acceptable for 'top N' scenarios. `RANK()` could also be used if ties should receive the same rank, but `ROW_NUMBER()` fulfills the requirement.
    *   `PARTITION BY c.CategoryName`: This clause divides the data into separate partitions for each category. The ranking will restart for each new category.
    *   `ORDER BY p.Price DESC`: Within each category partition, products will be ordered by their `Price` in descending order, so the most expensive product gets rank 1.

3.  **Main Query for Filtering**: After defining the CTE, the main query will select `CategoryName`, `ProductName`, and `Price` from the CTE.

4.  **Filter by Rank**: A `WHERE` clause will be applied to filter the results, keeping only those rows where the calculated `Rank` is less than or equal to 2. This effectively retrieves the top 2 most expensive products for each category.

This approach ensures that the ranking is performed correctly within each category independently, and the CTE enhances readability by separating the ranking logic from the final selection and filtering.

**Reasoning**:
Now that the reasoning for the SQL query is established, I will write the Python code to execute the SQL query using the `cursor` object, fetch the results, and then display them with column headers as per the subtask instructions.



````python
sql_query = '''
WITH RankedProducts AS (
    SELECT
        c.CategoryName,
        p.ProductName,
        p.Price,
        ROW_NUMBER() OVER (PARTITION BY c.CategoryName ORDER BY p.Price DESC) as Rank
    FROM
        Products p
    INNER JOIN
        Categories c ON p.CategoryID = c.CategoryID
)
SELECT
    CategoryName,
    ProductName,
    Price
FROM
    RankedProducts
WHERE
    Rank <= 2
ORDER BY
    CategoryName, Price DESC;
'''

cursor.execute(sql_query)
report_results = cursor.fetchall()

# Get column names for better display
column_names = [description[0] for description in cursor.description]

print("\nTop 2 Most Expensive Products per Category:")
print(column_names)
for row in report_results:
    print(row)
````

**Reasoning**:
Now that the mock Sakila database with customer and rental data is set up, I will construct the SQL query to identify customers with more than 10 rentals in the last 6 months, then execute it, fetch the results, and display them as per the subtask instructions.



````python
sql_query = '''
WITH LatestRentalDate AS (
    SELECT MAX(rental_date) AS max_rental_date FROM rental
),
CustomerRentalCounts AS (
    SELECT
        c.customer_id,
        c.first_name,
        c.last_name,
        c.email,
        COUNT(r.rental_id) AS rental_count
    FROM
        customer c
    JOIN
        rental r ON c.customer_id = r.customer_id
    JOIN
        LatestRentalDate lrd
    WHERE
        r.rental_date >= DATE(lrd.max_rental_date, '-6 months')
    GROUP BY
        c.customer_id, c.first_name, c.last_name, c.email
)
SELECT
    first_name || ' ' || last_name AS CustomerName,
    email,
    rental_count
FROM
    CustomerRentalCounts
WHERE
    rental_count > 10
ORDER BY
    rental_count DESC;
'''

sakila_cursor.execute(sql_query)
report_results = sakila_cursor.fetchall()

# Get column names for better display
column_names = [description[0] for description in sakila_cursor.description]

print("\nCustomers with More Than 10 Rentals in Last 6 Months:")
print(column_names)
for row in report_results:
    print(row)

# Close the connection to the mock Sakila database
sakila_conn.close()
print("\nConnection to mock Sakila database closed.")
````
